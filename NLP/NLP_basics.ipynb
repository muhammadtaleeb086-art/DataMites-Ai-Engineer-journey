#### Term frequency - Inverse Documents frequency

```python
import pandas as pd
import nltk
from sklearn.feature_extraction.text import TfidfVectorizer

corpus = [
  "Thor eating pizza, Loki is eating pizza , Ironman ate pizza already",
  "Apple is announcing new iphone tomorrow",
  "Tesla is announcing new model-3 tomorrow",
  "Google is annoucing new pixel-6 tomorrow",
  "Microsoft is annoucing new surface tomorrow",
  "Amazon is announcing new eco-dot tomorrow",
  "I am eating biryani and you are eating grapes"
]
```

```python
## Let's create the vectorizer and fit the corpus and transform them accordingly
from sklearn.feature_extraction.text import TfidfVectorizer
v = TfidfVectorizer()
v.fit(corpus)
transform_output = v.transform(corpus)
```

```python
## Let's print the vocabulary
print(v.vocabulary_)
```

```python
## Lets print ther idf of each word:

all_feature_names = v.get_feature_names_out()

for word in all_feature_names:

  ## Let's get the index in the vocabulary
  indx = v.vocabulary_.get(word)

  ## Get the score
  idf_score = v.idf_[indx]

  print(f"{word} : {idf_score}")
```

### Bags of Words

```python
corpus = [
  "Thor eating pizza, Loki is eating pizza , Ironman ate pizza already",
  "Apple is announcing new iphone tomorrow",
  "Tesla is announcing new model-3 tomorrow",
  "Google is annoucing new pixel-6 tomorrow",
  "Microsoft is annoucing new surface tomorrow",
  "Amazon is announcing new eco-dot tomorrow",
  "I am eating biryani and you are eating grapes"
]
```

```python
## Cleaning  the texts
import re
from nltk.corpus import stopwords ## Grammer removing
from nltk.stem.porter import PorterStemmer ## Stemming
from nltk.stem import WordNetLemmatizer  ## Lemmitization 
```

```python
import nltk
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('wordnet')

ps = PorterStemmer()
wordnet = WordNetLemmatizer()

## Store the original corpus data to iterate over
original_corpus_data = list(corpus)

## Clear thre corpus to store the clreaned reviews
corpus = []

for review_text in original_corpus_data:
  review = re.sub('[^a-zA-Z]', ' ',review_text)## removing unwanted words
  review = review.lower() ## lower the sentences
  review = review.split()
  review = [ps.stem(word) for word in review if not word in set(stopwords.words('english'))] #grammer 
  review = ' '.join(review) ## join the sentences
  corpus.append(review)
```

```python
## Creating the bag of Words model
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(max_features=1500) ## We can send this parameters (ngram_range=(1,3)
X = cv.fit_transform(corpus).toarray()
X
```
