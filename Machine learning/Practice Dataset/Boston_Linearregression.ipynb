```python
## Histogram
for column in df.columns:
  plt.figure(figsize=(6,4))
  sns.histplot(df[column],kde = True)
  plt.title(f'Distribution of {column}')
  plt.xlabel(column)
  plt.ylabel('Count')
  plt.show()
```

```python
## BoxPlot
for column in df.select_dtypes(include='number').columns:
    plt.figure(figsize=(6, 4))
    sns.boxplot(df[column])
    plt.title(f'Distribution of {column}')
    plt.xlabel(column)
    plt.ylabel('Count')
    plt.show()
```

```python
## Heatmap
plt.figure(figsize=(12, 8))

sns.heatmap(df.corr(), annot=True, cmap='coolwarm')

plt.title("Correlation Heatmap")
plt.show()
```

```python
## Scatter plot
for column in df.columns:
    if column != 'medv':
        plt.figure(figsize=(6, 4))
        sns.scatterplot(x=df[column], y=df['medv'])
        plt.xlabel(column)
        plt.ylabel('medv')
        plt.title(f'{column} vs medv')
        plt.show()
```

```python
X = df[['crim', 'zn', 'indus', 'chas', 'nox', 'rm', 'age', 'dis',
        'rad', 'tax', 'ptratio', 'black', 'lstat']]

y = df['medv']
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2)
```

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
```

```python
model.fit(X_train,y_train)
```

```python
model.score(X_test,y_test)
```

```python
model.predict(X_test)
```

```python
y_pred = model.predict(X_test)
```

```python
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error
r2Score = r2_score(y_test,y_pred)
mean_squared_error = mean_squared_error(y_test, y_pred)
mean_absolute_error = mean_absolute_error(y_test, y_pred)
```

```python
print(f" R2 score: {r2Score},\n Mean_squared_error: {mean_squared_error},\n mean_absolute_error: {mean_absolute_error}")
```

```python
## Actual vs Predicted
comparison = pd.DataFrame({
    'Actual': y_test,
    'Predicted': y_pred
})

comparison.head(10)
```

```python
plt.figure(figsize=(7,5))

plt.scatter(y_test, y_pred)
plt.xlabel("Actual MEDV")
plt.ylabel("Predicted MEDV")
plt.title("Actual vs Predicted")

plt.show()
```

```python
residuals = y_test - y_pred
residuals
```

```python
plt.figure(figsize=(7,5))

plt.scatter(y_pred, residuals)
plt.axhline(y=0)
plt.xlabel("Predicted MEDV")
plt.ylabel("Residuals")
plt.title("Residual Plot")

plt.show()
```
