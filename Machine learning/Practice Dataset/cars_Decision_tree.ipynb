```python
## how to convert categorical data into numerical data
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
```

```python
df['buying'] = le.fit_transform(df['buying'])
df['maint'] = le.fit_transform(df['maint'])
df['doors'] = le.fit_transform(df['doors'])
df['persons'] = le.fit_transform(df['persons'])
df['lug_boot'] = le.fit_transform(df['lug_boot'])
df['safety'] = le.fit_transform(df['safety'])
df['car'] = le.fit_transform(df['car'])
```

```python
## EDA Graph
## CountPlot
for col in df.columns:
    plt.figure(figsize=(7, 4))
    sns.countplot(data=df, x=col)
    plt.title(f'Distribution of {col}')
    plt.xlabel(col)
    plt.ylabel('Count')
    plt.show()
```

```python
plt.figure(figsize=(7, 4))
sns.countplot(data=df, x='car')
plt.title('Distribution of Car Evaluation')
plt.xlabel('Car Evaluation')
plt.ylabel('Count')
plt.show()
```

```python
for col in  df.columns:
    plt.figure(figsize=(8, 5))
    sns.countplot(data=df, x=col, hue='car')
    plt.title(f'{col} vs Car Evaluation')
    plt.xlabel(col)
    plt.ylabel('Count')
    plt.xticks(rotation=0)
    plt.legend(title='Car')
    plt.show()
```

```python
## Feature Selection 
X = df[['buying', 'maint', 'doors', 'persons', 'lug_boot', 'safety']]
y = df['car']
```

```python
## train test split
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2)
```

```python
from sklearn.tree import DecisionTreeClassifier
model = DecisionTreeClassifier()
```

```python
model.fit(X_train,y_train)
```

```python
model.score(X_test,y_test)
```

```python
model.predict(X_test)
```

```python
y_pred = model.predict(X_test)
```

```python
from sklearn.metrics import classification_report , confusion_matrix

print("Classification Report")
cr = classification_report(y_test,y_pred)
print(cr)

print("-"*40)

print("Confusion Matrix")
cm = confusion_matrix(y_test,y_pred)
print(cm)
```

```python
plt.figure(figsize=(6, 4))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues')
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.title('Confusion Matrix')
plt.show()
```

```python
importance = pd.Series(
    model.feature_importances_,
    index=X.columns
).sort_values(ascending=False)

print(importance)
```

```python
plt.figure(figsize=(8, 5))
importance.plot(kind='bar')
plt.title('Feature Importance')
plt.xlabel('Features')
plt.ylabel('Importance')
plt.show()
```
