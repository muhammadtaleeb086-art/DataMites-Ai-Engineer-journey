```python
## Loc means location. It is used to get the rows and columns with particular condition. 

## iloc means index location. It is used to get the rows and columns with particular condition. 
```

```python
data.loc[data['Age']==80] # loc is used to get the rows and columns with particular condition. It will give the rows where age is 80.
```

```python
data.loc[(data['Sex']=='male') & (data['Age']>40)] # It will give the rows where sex is male and age is greater than

## In this above code we can also check >,<,>=,<=,==,!= conditions. We can also use or condition using | operator.
```

```python
## age > 22 , fare > 10 , parch >1
data.loc[(data['Age']>22) & (data['Fare']>10) & (data['Parch']>1) & (data['Sex']=='female')] ## For femaele
```

```python
data.loc[(data['Age']>22) & (data['Fare']>10) & (data['Parch']>1) & (data['Sex']=='male')] ## for male
```

```python
# number of females in first class
data.loc[(data["Sex"]=="female") & (data["Pclass"]==1)]
```

```python
data.iloc[np.where(data['Age']==80)] ## iloc is used to get the rows and columns with particular condition. It will give the rows where age is 80.
```

```python
data.iloc[np.where((data['Sex']=='male') & (data['Survived']==1))] ## It will give the rows where sex is male and survived is 1. We can also use or condition using | operator.
```

```python
data.iloc[np.where((data['Fare']>10) & (data['Age']>50) & (data['Parch']==1) & (data['Sex']=='male'))]
```

```python
data.iloc[np.where((data['Fare']>10) & (data['Age']>50) & (data['Parch']==1) & (data['Sex']=='female'))]
```
