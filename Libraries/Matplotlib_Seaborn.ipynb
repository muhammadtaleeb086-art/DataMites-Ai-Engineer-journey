```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np
```

```python
sns.get_dataset_names()
```

```python
## Loading dataset
df = sns.load_dataset('iris')
```

```python
df.species.unique()
```

```python
df.species.value_counts()
```

```python
df.columns
```

```python
sns.lineplot(x='sepal_length',y = 'sepal_width',data=df)
plt.show()
```

```python
sns.lineplot(x='sepal_length',y = 'sepal_width',data=df,hue='species',ci=None)
plt.show()
```

```python
## Scatter Plot
sns.scatterplot(x='sepal_length',y='sepal_width',data = df , hue = 'species')
plt.title("Scatter Plot")
plt.show()
```

```python
## Bar Plot (categorical estimate plot)
## A barplot is basically used to aggreate the categorical data according to some methods any by default it 

sns.barplot(x='species',y='sepal_length',data=df)
plt.show()
```

```python
sns.countplot(x='species',data=df)
plt.show()
```

```python
sns.boxplot(x='sepal_width',data=df)
plt.show()
```

```python
sns.boxplot(x='species',y='sepal_width',data=df)
plt.show()
```

```python
sns.violinplot(x='species',y='sepal_width',data=df)
plt.show()
```

```python
sns.stripplot(x='species',y='petal_width',data=df)
plt.show()
```

```python
sns.histplot(x='petal_width',data=df)
plt.show()
```

```python
sns.histplot(x='petal_width',data=df,hue='species',kde=True)
plt.show()
```

```python
 
sns.distplot(df['petal_width'],color='r')
plt.show()
```

```python
sns.pairplot(data=df,hue='species')
plt.show()
```

```python
sns.catplot(x='species',y='sepal_length',data=df,kind = 'swarm')
plt.show()
```

```python
sns.catplot(x='petal_length',y='species',data=df)
plt.show()
```

```python
df.drop('species',axis=1,inplace=True)
tc = df.corr()
tc
```

```python
sns.heatmap(tc,annot=True)
plt.show()
```
