```python
## 8. Find the largest of three numbers.
a = int(input("Enter a number (A):"))
b = int(input("Enter a number (B):"))
c = int(input("Enter a number (C):"))

if a >= b and a >= c :
  print(f"A={a} is the largest number.")
elif b >= a and b >= c:
  print(f"B={b} is the largest number.")
else :
  print(f"C={c} is largest number.")
```

```python
## 9. Check whether a year is a leap year.

year = int(input("Enter a year :"))

if year % 400 == 0 :
  print(f"{year} is  a leap year ")
elif year % 100 == 0 :
  print(f"{year} is not a leap year.")
elif year % 4 == 0 :
  print(f"{year} is a leap year")
else :
  print(f"{year} is not a leap year")
```

```python
## 10. Check if a character is a vowel or consonant.

vowels = ['a', 'e', 'i', 'o', 'u']

letter = input("Enter a letter: ").lower()

if len(letter) != 1 or not letter.isalpha():
    print("Invalid input! Please enter a single alphabet letter.")
elif letter in vowels:
    print(f"{letter} is a vowel.")
else:
    print(f"{letter} is a consonant.")

```

```python
# 11. Build a simple calculator using `+`, `-`, `*`, and `/`.

num1 = float(input("Enter a number : "))
num2 = float(input("Enter a number : "))

print("---Calculator---")
print("1.Addition(+)")
print("2.Subtraction(-)")
print("3.Multiplication(*)")
print("4.Division(/)")

choice =  int(input("enter which operation you want perform (1/2/3/4) :"))

if choice == 1 :
  print(f"Addition : {num1 + num2}")
elif choice == 2 :
  print(f"Subtraction : {num1 - num2}")
elif choice == 3:
  print(f"Multiplication : {num1 * num2}")
elif choice == 4 :
  if num2 == 0 :
    print("Cannot divide by zero.")
  else :
    print(f"Division : {num1 / num2}")
else :
  print("Invalid choice.")

```

```python
# 21. Find the frequency of each character.
text = input("Enter a string: ")

frequency = {}

for char in text:
    if char in frequency:
        frequency[char] += 1
    else:
        frequency[char] = 1

print(frequency)

```

```python
# 22. Convert every alternate letter to uppercase.
text = input("Enter a text :")
result = ""
for i in range(len(text)):
  if i % 2 == 0 :
    result += text[i].upper()
  else :
    result += text[i].lower()

print(result)
```

```python
## 27.Write a function `counting(lst, num)` that counts how many times `num` appears in `lst` (you already have this — extend it to also work when `num` isn't found).

def counting(lst,num):
    count = 0
    for i in lst:
        if i == num:
            count += 1
    if count == 0 :
            print("number is not found in list.")
    return count 
```

```python
counting([2,3,4,56,3,2,56,2],99)
```

```python
## 30.Use `map()` with a lambda to convert a list of Celsius temperatures to Fahrenheit.
celsius = [0, 10, 20, 30, 37]

fahrenheit = list(map(lambda c: (c * 9/5) + 32, celsius))
print(fahrenheit)
```
