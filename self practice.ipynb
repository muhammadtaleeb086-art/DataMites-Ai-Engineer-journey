```python
## 9. Check whether a year is a leap year.

year = int(input("Enter a year :"))

if year % 400 == 0 :
  print(f"{year} is  a leap year ")
elif year % 100 == 0 :
  print(f"{year} is not a leap year.")
elif year % 4 == 0 :
  print(f"{year} is a leap year")
else :
  print(f"{year} is not a leap year")
```

```python
## 10. Check if a character is a vowel or consonant.

vowels = ['a', 'e', 'i', 'o', 'u']

letter = input("Enter a letter: ").lower()

if len(letter) != 1 or not letter.isalpha():
    print("Invalid input! Please enter a single alphabet letter.")
elif letter in vowels:
    print(f"{letter} is a vowel.")
else:
    print(f"{letter} is a consonant.")

```

```python
## 27.Write a function `counting(lst, num)` that counts how many times `num` appears in `lst` (you already have this — extend it to also work when `num` isn't found).

def counting(lst,num):
    count = 0
    for i in lst:
        if i == num:
            count += 1
    if count == 0 :
            print("number is not found in list.")
    return count 
```

```python
## 31. Find the largest element in a list. 
lst = [12,34,56,78,9,0,1990]
# max(lst)
def maximum(lst):
  if len(lst)==0:
    return "no element found"
    exit()
  maximum = lst[0]
  for i in lst:
    if i >= maximum :
      maximum = i
  return maximum

maximum(lst)
  
```

```python
## 32.Find the second largest element.
lst = [12, 34, 56, 78, 9, 0, 1990]

def second_largest(lst):
    if len(lst) < 2:
        return "List must contain at least 2 elements."

    largest = second = float('-inf')

    for num in lst:
        if num > largest:
            second = largest
            largest = num
        elif num > second and num != largest:
            second = num

    if second == float('-inf'):
        return "Second largest element not found."

    return second

print(second_largest(lst))
```

```python
## 40. Sort a list without using `sort()` (implement Bubble Sort).
lst = [95, 62, 32, 1, 23, 252, 45, 26, 96, 84, 75, 85, 66, 354, 200]

n = len(lst)

for i in range(n):
    for j in range(n - i - 1):
        if lst[j] > lst[j + 1]:
            lst[j], lst[j + 1] = lst[j + 1], lst[j]

print(lst)
```
