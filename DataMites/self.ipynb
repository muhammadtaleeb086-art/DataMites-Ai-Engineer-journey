# Identifiers

<p>
A Python identifier is the name used to identify variables, functions, classes, modules, and other objects in a Python program. Identifiers help programmers refer to different elements of the code in a meaningful way. They can contain letters (A-Z, a-z), digits (0-9), and underscores (_), but they cannot start with a digit. Python identifiers are case-sensitive, which means `name`, `Name`, and `NAME` are considered different identifiers. Special characters such as `@`, `#`, and `$` are not allowed in identifiers. Additionally, Python keywords like `if`, `for`, and `while` cannot be used as identifiers because they have predefined meanings in the language. Using clear and descriptive identifiers improves code readability and makes programs easier to understand and maintain.

Multi-line statements are commonly used for long mathematical expressions, large lists, dictionaries, or complex function calls. They help keep code neat and easier to maintain.
</p>

# Indexing


Python **indexing** is the process of accessing individual elements of a sequence, such as a string, list, tuple, or array, using their position number called an **index**. In Python, indexing starts from **0**, so the first element is at index `0`, the second at index `1`, and so on. Python also supports **negative indexing**, where `-1` refers to the last element of the sequence. Indexing allows programmers to retrieve, modify (for mutable objects like lists), or inspect specific elements efficiently. It is one of the most fundamental concepts used in data manipulation and programming. Using an invalid index results in an **IndexError**.


```python
txt = "My name is Taleeb "

# Positive Indexing 
txt[9]
txt[11]
```

```python
# Negative Indexing 
txt[-2]
```

### Slicing 
Slicing is the process of extracting a portion (subsequence) of a sequence, such as a string, list, or tuple, by specifying a start index, a stop index, and an optional step.

sequence[start:stop:step]

- start: Index where slicing begins (inclusive).
- stop: Index where slicing ends (exclusive).
- step: Number of positions to move between elements (optional, default is 1).

```python
txt[11:17:1]
```

```python
txt[-6::1]
```

```python
txt= "hritik is the trainer"
```

```python
txt[-9:-12:-1]
```

```python
txt[-16::-1]
```

```python
nums = [0, 1, 2, 3, 4, 5]

print(nums[1:4])    # [1, 2, 3]
print(nums[:3])     # [0, 1, 2]
print(nums[3:])     # [3, 4, 5]
print(nums[::2])    # [0, 2, 4]
print(nums[::-1])   # [5, 4, 3, 2, 1, 0]
```

### Autotype Casting
Auto type casting, also called implicit type casting, is the automatic conversion of a smaller data type into a larger compatible data type by Python during an operation. This happens without using any conversion function.

```python
x = 10      # int
y = 2.5     # float

result = x + y
print(result)        # 12.5
print(type(result))  # <class 'float'>
```

## DateTime

The datetime module is a Python module that provides classes for handling dates, times, date and time combinations, time intervals, and time zone information.

Main classes in the datetime module
- date: Represents a date (year, month, day).
- time: Represents a time (hour, minute, second, microsecond).
- datetime: Represents both a date and a time.
- timedelta: Represents the difference (duration) between two dates or times.
- timezone: Represents time zone information.
