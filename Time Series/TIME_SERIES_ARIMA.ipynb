```python
 import statsmodels.api as sm
 from statsmodels.tsa.stattools import adfuller
 from statsmodels.tsa.arima.model import ARIMA
```

```python
ts=sm.datasets.sunspots.load_pandas().data["SUNACTIVITY"]
```

```python
ts.head()
```

```python
#if it is above 0.05 mean is not constant/non stationery
#if it is below 0.05 mean is constant/ stationery
#in this case the mean is above a small margin so it is okay we can use it
adfuller(ts)[1]
```

```python
arima=ARIMA(ts,order=(3,2,4))
model=arima.fit()
```

```python
model.aic
```

```python
model.forecast(steps=5)
```
